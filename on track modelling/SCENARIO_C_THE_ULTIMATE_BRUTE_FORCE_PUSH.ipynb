```python
# ============================================================
# SCENARIO C - THE ULTIMATE BRUTE-FORCE PUSH (>85%)
# KLASIFIKASI CITRA OBAT TABLET
# ============================================================

import os
import glob
import numpy as np
import tensorflow as tf

from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.losses import CategoricalCrossentropy

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
from google.colab import drive

# ============================================================
# 1. GOOGLE DRIVE (Sistem Deteksi Otomatis)
# ============================================================
print("🔌 Connecting Google Drive...")
try:
    drive.mount("/content/drive")
except:
    print("✅ Drive already mounted")

# ============================================================
# 2. PATH DATASET (Sesuai Konfigurasi Sukses Sebelumnya)
# ============================================================
BASE_DIR = "/content/drive/MyDrive/SKRIPSI-ALWI-COLAB"
BASELINE_DIR = os.path.join(BASE_DIR, "baseline_dataset")
MODELS_DIR = os.path.join(BASE_DIR, "models")
OUTPUTS_DIR = os.path.join(BASE_DIR, "outputs", "scenario_c_final")

os.makedirs(MODELS_DIR, exist_ok=True)
os.makedirs(OUTPUTS_DIR, exist_ok=True)
MODEL_SAVE_PATH = os.path.join(MODELS_DIR, "scenario_c_bruteforce.h5")

train_path = os.path.join(BASELINE_DIR, "train")
valid_path = os.path.join(BASELINE_DIR, "valid")
test_path = os.path.join(BASELINE_DIR, "test")

for folder in [train_path, valid_path, test_path]:
    if not os.path.exists(folder):
        raise Exception(f"❌ FATAL ERROR: Folder tidak ditemukan -> {folder}")
print("✅ Direktori Sistem Terverifikasi")

# ============================================================
# 3. TARGET CLASS
# ============================================================
TARGET_CLASSES = [
    "cataflam_50_mg", "cetirizin_10_mg", "pantoprazol_sandoz_40_mg",
    "c_vitamin_teva_500_mg", "aspirin_ultra_500_mg",
    "algoflex_forte_dolo_400_mg", "merckformin_xr_1000_mg"
]

print("\n🔥 MENYIAPKAN AMUNISI DATA...")
IMG_SIZE = (224, 224)
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode="nearest"
)
valid_test_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

train_generator = train_datagen.flow_from_directory(
    train_path, target_size=IMG_SIZE, batch_size=BATCH_SIZE, classes=TARGET_CLASSES, class_mode="categorical", shuffle=True)
valid_generator = valid_test_datagen.flow_from_directory(
    valid_path, target_size=IMG_SIZE, batch_size=BATCH_SIZE, classes=TARGET_CLASSES, class_mode="categorical", shuffle=False)
test_generator = valid_test_datagen.flow_from_directory(
    test_path, target_size=IMG_SIZE, batch_size=BATCH_SIZE, classes=TARGET_CLASSES, class_mode="categorical", shuffle=False)

# ============================================================
# 4. ARSITEKTUR MODEL
# ============================================================
base_model = EfficientNetB0(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation='relu')(x)
x = BatchNormalization()(x)
x = Dropout(0.5)(x) # Dropout dinaikkan agar model tidak menghafal
predictions = Dense(len(TARGET_CLASSES), activation='softmax')(x)

model_brute = Model(inputs=base_model.input, outputs=predictions)

checkpoint = ModelCheckpoint(filepath=MODEL_SAVE_PATH, monitor='val_accuracy', save_best_only=True, verbose=1)
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

# 💣 Trik Brute-Force 1: Label Smoothing
loss_fn = CategoricalCrossentropy(label_smoothing=0.15)

# ============================================================
# 5. TRAINING TAHAP 1 (WARM UP)
# ============================================================
model_brute.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss=loss_fn, metrics=['accuracy'])

print("\n🚀 TAHAP 1: PEMANASAN MESIN...")
history_step1 = model_brute.fit(
    train_generator,
    epochs=15,
    validation_data=valid_generator,
    callbacks=[checkpoint, early_stop],
    verbose=1
)

# ============================================================
# 6. TRAINING TAHAP 2 (100% UNFREEZE BRUTE-FORCE)
# ============================================================
# 💣 Trik Brute-Force 2: Buka Kunci Seluruh Otak Model!
base_model.trainable = True

# Menggunakan learning rate yang sangat kecil agar otak lamanya tidak hancur seketika
model_brute.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5), loss=loss_fn, metrics=['accuracy'])
lr_scheduler = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7, verbose=1)

print("\n🚀 TAHAP 2: 100% SYSTEM OVERRIDE (Membongkar seluruh layer EfficientNet)...")
history_fine = model_brute.fit(
    train_generator,
    epochs=35, # Waktu training diperpanjang
    validation_data=valid_generator,
    callbacks=[checkpoint, early_stop, lr_scheduler],
    verbose=1
)

# ============================================================
# 7. EVALUASI SUPER AKHIR
# ============================================================
print("\n🔍 MENGHITUNG KEMENANGAN PADA DATA UJI...")
best_model = tf.keras.models.load_model(MODEL_SAVE_PATH)
test_loss, test_acc = best_model.evaluate(test_generator, verbose=0)

print("="*50)
print(f"🌟 AKURASI FINAL BRUTE-FORCE : {test_acc * 100:.2f}% 🌟")
print("="*50)

test_generator.reset()
predictions = best_model.predict(test_generator, verbose=1)
predicted_classes = np.argmax(predictions, axis=1)
true_classes = test_generator.classes

report = classification_report(true_classes, predicted_classes, target_names=TARGET_CLASSES)
print(report)

plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix(true_classes, predicted_classes), annot=True, fmt='d', cmap='viridis', xticklabels=TARGET_CLASSES, yticklabels=TARGET_CLASSES)
plt.title(f'Confusion Matrix - Brute Force ({test_acc * 100:.2f}%)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.ylabel('Aktual (Kebenaran)')
plt.xlabel('Prediksi Model')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUTS_DIR, 'bruteforce_confusion_matrix.png'), dpi=300)
plt.show()
```
